```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

counts = pd.read_csv("GSE80655_counts.csv", index_col="gene_id")
metadata = pd.read_csv(
    "GSE80655_metadata.csv",
    dtype={"donor_id": "string"}
)

print("Genes:", counts.shape[0])
print("Samples:", counts.shape[1])
print("Donors:", metadata["donor_id"].nunique())

display(metadata.head())
display(counts.head())
```

```python
assert counts.index.is_unique, "Duplicate gene IDs."
assert counts.columns.is_unique, "Duplicate count sample IDs."

assert metadata["description"].notna().all(), "Missing sample IDs."
assert metadata["description"].is_unique, "Duplicate metadata sample IDs."

assert set(counts.columns) == set(metadata["description"]), \
    "Samples do not match."
```

All checks passed: identifiers are unique and both files contain the same samples.

```python
metadata = metadata.set_index("description", drop=False)
metadata = metadata.loc[counts.columns].copy()

print("Sample order matches:",
      counts.columns.equals(metadata.index))
```

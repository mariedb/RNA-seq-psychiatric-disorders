```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import datasets
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.model_selection import train_test_split

# Set the style for our visualizations
sns.set_style("whitegrid")
sns.set_palette("deep")

%matplotlib inline

counts = pd.read_csv("GSE80655_counts.csv")
metadata = pd.read_csv("GSE80655_metadata.csv")
```

```python
display(counts.head())
display(metadata.head())
```

```python
print("Counts:", counts.shape)
print("Metadata:", metadata.shape)
```

```python
# Check missing values
print("Missing values in counts:", counts.isnull().sum())
print("Missing values in metadata:", metadata.isnull().sum())
```

```python
missing_ph = metadata[metadata["brain_pH"].isna()]
print(missing_ph[["donor_id", "brain_region", "diagnosis", "age", "PMI"]])
```

```python
# Check if sample IDs match
count_samples = counts.columns[1:]
metadata_samples = metadata["description"]

print("Samples in counts:", len(count_samples))
print("Samples in metadata:", len(metadata_samples))
print("Matching samples:", sum(count_samples.isin(metadata_samples)))

print("Unmatched samples:", sum(~count_samples.isin(metadata_samples)))
```

```python
# Calculate total counts per sample
total_counts = counts.iloc[:, 1:].sum()

# Display summary statistics
total_counts.describe()
```

```python
# Select only gene expression columns
gene_counts = counts.iloc[:, 1:]

# Check dimensions
print("Original counts:", counts.shape)
print("Gene counts:", gene_counts.shape)

# Count in how many samples each gene is expressed
expressed_samples = (gene_counts > 0).sum(axis=1)

# Check number of genes
print("Total genes:", len(expressed_samples))
print("Genes with no expression:", (expressed_samples == 0).sum())
print("Genes expressed in at least one sample:", (expressed_samples > 0).sum())

# Check unique gene IDs
print("Unique gene IDs:", counts["gene_id"].nunique())
```

```python
# Remove genes with zero counts in all samples
filtered_counts = counts[gene_counts.sum(axis=1) > 0].copy()

print("Genes before filtering:", counts.shape[0])
print("Genes after filtering:", filtered_counts.shape[0])
```
